# Seed-paired statistical analysis of the observation-level calibration benchmark

This notebook analyses every completed static-PCA and MWPCA method/configuration while treating seed—not detector row—as the independent experimental unit. Its primary outcome is restricted detection delay at the Phase-II horizon, so undetected episodes remain in the analysis. Comparisons are paired within seed and exact drift condition. Because the design uses one Phase-II episode per seed, results should be interpreted as exploratory screening estimates.


## Configuration


```python
from pathlib import Path
import os

folder_candidate = Path.home() / "Documents" / "Stats Thesis notebooks and results" / "02_observation_level_calibration_extension"
PACKAGE_ROOT = next((root for root in (Path.cwd(), folder_candidate)
                     if (root / "seed_paired_method_analysis.py").is_file()), Path.cwd())
BASE_ROOT = PACKAGE_ROOT
RESULT_ROOT = (
    BASE_ROOT / "outputs/observation_level_calibration_static"
)
ANALYSIS_ROOT = RESULT_ROOT / "statistical_analysis"
BOOTSTRAP_REPS = int(os.environ.get("METHOD_ANALYSIS_BOOTSTRAP_REPS", "2000"))

namespace = {}
script = PACKAGE_ROOT / "seed_paired_method_analysis.py"
exec(compile(script.read_text(), str(script), "exec"), namespace)
analyse = namespace["analyse"]
print(f"result root: {RESULT_ROOT}")
print(f"analysis output: {ANALYSIS_ROOT}")
```

## Run the complete analysis


```python
ANALYSIS = analyse(
    RESULT_ROOT,
    ANALYSIS_ROOT,
    bootstrap_reps=BOOTSTRAP_REPS,
)
display(ANALYSIS["summaries"].head(30))
display(ANALYSIS["comparisons"].head(30))
display(ANALYSIS["winners"].head(30))
display(ANALYSIS["associations"].head(30))
```

## Outputs

The analysis writes seed-level outcomes, per-method condition summaries, paired restricted-delay comparisons, exact paired detection comparisons, Benjamini--Hochberg adjusted values, condition rankings, winner counts, and held-out-ARL0/delay associations. A negative `mean_restricted_delay_difference_a_minus_b` means configuration A was faster than configuration B. Rankings use the mean across seeds for each detector--configuration pair; they never select an individual minimum run.
